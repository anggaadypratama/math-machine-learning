## 🎯 Section 2: Limit - Fondasi Kalkulus

### Apa itu Limit?

Limit adalah konsep fundamental dalam kalkulus yang menggambarkan nilai yang "didekati" oleh suatu fungsi ketika variabel input mendekati nilai tertentu.

Bayangkan kamu lagi jalan mendekati pintu rumah. Limit itu seperti "di mana kamu akan sampai" kalau kamu terus berjalan ke arah pintu, meski mungkin kamu nggak pernah benar-benar sampai di pintu!

### Notasi Limit

$$\lim_{x \to a} f(x) = L$$

Dibaca: "Limit dari f(x) ketika x mendekati a adalah L"

### Jenis-jenis Limit

#### 1. Limit Finite
$$\lim_{x \to 2} (3x + 1) = 7$$

#### 2. Limit Tak Hingga
$$\lim_{x \to \infty} \frac{1}{x} = 0$$

#### 3. Limit Kiri dan Kanan
$$\lim_{x \to a^-} f(x) \text{ dan } \lim_{x \to a^+} f(x)$$

### Teorema Limit Penting

1. **Limit Penjumlahan**: $\lim_{x \to a} [f(x) + g(x)] = \lim_{x \to a} f(x) + \lim_{x \to a} g(x)$
2. **Limit Perkalian**: $\lim_{x \to a} [f(x) \cdot g(x)] = \lim_{x \to a} f(x) \cdot \lim_{x \to a} g(x)$
3. **Limit Pembagian**: $\lim_{x \to a} \frac{f(x)}{g(x)} = \frac{\lim_{x \to a} f(x)}{\lim_{x \to a} g(x)}$ (jika pembagi ≠ 0)

### Bentuk Tak Tentu

- $\frac{0}{0}$: Gunakan L'Hôpital's Rule atau faktorisasi
- $\frac{\infty}{\infty}$: Gunakan L'Hôpital's Rule  
- $0 \cdot \infty$: Ubah bentuk
- $\infty - \infty$: Faktorisasi

*Referensi: [Khan Academy - Introduction to Limits](https://www.khanacademy.org/math/calculus-1/cs1-limits-and-continuity), [Paul's Online Math Notes - Limits](https://tutorial.math.lamar.edu/Classes/CalcI/TheLimit.aspx)*

````python
# Implementasi dan Visualisasi Limit
print("🎯 KONSEP LIMIT DAN IMPLEMENTASI")
print("=" * 50)

# Definisikan simbolik variable
x = sp.Symbol('x')

# Contoh 1: Limit sederhana
print("📚 CONTOH 1: Limit Sederhana")
f1 = 3*x + 2
limit1 = sp.limit(f1, x, 2)
print(f"f(x) = {f1}")
print(f"lim(x→2) f(x) = {limit1}")

# Verifikasi numerik
x_vals = np.array([1.9, 1.99, 1.999, 1.9999, 2.0001, 2.001, 2.01, 2.1])
f_vals = 3*x_vals + 2
print(f"\nVerifikasi numerik:")
for i, (x_val, f_val) in enumerate(zip(x_vals, f_vals)):
    print(f"f({x_val}) = {f_val}")

# Contoh 2: Limit dengan bentuk tak tentu 0/0
print(f"\n📚 CONTOH 2: Limit Bentuk Tak Tentu (0/0)")
f2 = (x**2 - 4)/(x - 2)
limit2 = sp.limit(f2, x, 2)
print(f"f(x) = {f2}")
print(f"lim(x→2) f(x) = {limit2}")

# Faktorisasi untuk pemahaman
f2_simplified = sp.simplify(f2)
print(f"Setelah disederhanakan: f(x) = {f2_simplified} (untuk x ≠ 2)")

# Contoh 3: Limit tak hingga
print(f"\n📚 CONTOH 3: Limit Tak Hingga")
f3 = 1/x
limit3_pos = sp.limit(f3, x, sp.oo)
limit3_neg = sp.limit(f3, x, -sp.oo)
print(f"f(x) = {f3}")
print(f"lim(x→∞) f(x) = {limit3_pos}")
print(f"lim(x→-∞) f(x) = {limit3_neg}")

# Visualisasi limit
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(15, 12))

# Plot 1: Limit sederhana
x_range = np.linspace(-1, 5, 1000)
y1 = 3*x_range + 2

ax1.plot(x_range, y1, 'b-', linewidth=2, label='f(x) = 3x + 2')
ax1.plot(2, 8, 'ro', markersize=10, label='lim(x→2) = 8')
ax1.axhline(y=8, color='r', linestyle='--', alpha=0.7)
ax1.axvline(x=2, color='r', linestyle='--', alpha=0.7)
ax1.set_xlim(-1, 5)
ax1.set_ylim(-1, 15)
ax1.set_title('Limit Sederhana: lim(x→2) (3x+2) = 8', fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.legend()

# Plot 2: Limit dengan hole (bentuk tak tentu)
x_range2 = np.linspace(-1, 5, 1000)
x_range2 = x_range2[x_range2 != 2]  # Remove x=2
y2 = (x_range2**2 - 4)/(x_range2 - 2)

ax2.plot(x_range2, y2, 'b-', linewidth=2, label='f(x) = (x²-4)/(x-2)')
ax2.plot(2, 4, 'ro', markersize=10, label='lim(x→2) = 4')
ax2.plot(2, 4, 'wo', markersize=8, markeredgecolor='red', markeredgewidth=2, label='Hole at x=2')
ax2.axhline(y=4, color='r', linestyle='--', alpha=0.7)
ax2.axvline(x=2, color='r', linestyle='--', alpha=0.7)
ax2.set_xlim(0, 4)
ax2.set_ylim(2, 6)
ax2.set_title('Limit dengan Hole: lim(x→2) (x²-4)/(x-2) = 4', fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.legend()

# Plot 3: Limit tak hingga
x_range3 = np.linspace(-10, 10, 1000)
x_range3 = x_range3[x_range3 != 0]  # Remove x=0
y3 = 1/x_range3

ax3.plot(x_range3, y3, 'b-', linewidth=2, label='f(x) = 1/x')
ax3.axhline(y=0, color='r', linestyle='--', alpha=0.7, label='y = 0 (horizontal asymptote)')
ax3.axvline(x=0, color='g', linestyle='--', alpha=0.7, label='x = 0 (vertical asymptote)')
ax3.set_xlim(-10, 10)
ax3.set_ylim(-5, 5)
ax3.set_title('Limit Tak Hingga: lim(x→∞) 1/x = 0', fontweight='bold')
ax3.grid(True, alpha=0.3)
ax3.legend()

# Plot 4: Pendekatan numerik ke limit
x_approaches = np.array([1.9, 1.99, 1.999, 1.9999, 2.0001, 2.001, 2.01, 2.1])
y_approaches = 3*x_approaches + 2

ax4.plot(x_approaches, y_approaches, 'bo-', markersize=8, linewidth=2, label='Nilai f(x) mendekati 2')
ax4.axhline(y=8, color='r', linestyle='--', alpha=0.7, label='Limit = 8')
ax4.axvline(x=2, color='r', linestyle='--', alpha=0.7, label='x = 2')

# Annotate points
for i, (x_val, y_val) in enumerate(zip(x_approaches, y_approaches)):
    if i < 4:  # Left approach
        ax4.annotate(f'({x_val:.3f}, {y_val:.1f})', (x_val, y_val), 
                    xytext=(-20, 10), textcoords='offset points', fontsize=8,
                    arrowprops=dict(arrowstyle='->', color='blue', alpha=0.7))
    else:  # Right approach
        ax4.annotate(f'({x_val:.3f}, {y_val:.1f})', (x_val, y_val), 
                    xytext=(20, -15), textcoords='offset points', fontsize=8,
                    arrowprops=dict(arrowstyle='->', color='blue', alpha=0.7))

ax4.set_xlim(1.8, 2.2)
ax4.set_ylim(7.5, 8.5)
ax4.set_title('Pendekatan Numerik ke Limit', fontweight='bold')
ax4.grid(True, alpha=0.3)
ax4.legend()

plt.tight_layout()
plt.show()

# L'Hôpital's Rule untuk bentuk tak tentu
print(f"\n📐 L'HÔPITAL'S RULE")
print("Untuk bentuk tak tentu 0/0 atau ∞/∞:")

# Contoh L'Hôpital's Rule
f_num = sp.sin(x)
f_den = x
print(f"\nContoh: lim(x→0) sin(x)/x")

# Cek bentuk tak tentu
at_zero_num = f_num.subs(x, 0)
at_zero_den = f_den.subs(x, 0)
print(f"sin(0) = {at_zero_num}, 0 = {at_zero_den} → bentuk 0/0")

# Aplikasi L'Hôpital's Rule
f_num_prime = sp.diff(f_num, x)
f_den_prime = sp.diff(f_den, x)
print(f"Turunan pembilang: d/dx[sin(x)] = {f_num_prime}")
print(f"Turunan penyebut: d/dx[x] = {f_den_prime}")

lhopital_result = sp.limit(f_num_prime/f_den_prime, x, 0)
print(f"lim(x→0) cos(x)/1 = {lhopital_result}")

# Verifikasi dengan limit asli
original_limit = sp.limit(f_num/f_den, x, 0)
print(f"Verifikasi: lim(x→0) sin(x)/x = {original_limit}")

# Continuity check
print(f"\n🔗 KONTINUITAS")
print("Fungsi f(x) kontinu di x=a jika:")
print("1. f(a) ada (terdefinisi)")
print("2. lim(x→a) f(x) ada") 
print("3. lim(x→a) f(x) = f(a)")

# Contoh fungsi kontinu dan diskontinu
x_cont = np.linspace(-3, 3, 1000)

# Fungsi kontinu
y_cont = x_cont**2

# Fungsi diskontinu (step function)
y_discont = np.piecewise(x_cont, 
                        [x_cont < 0, x_cont >= 0], 
                        [lambda x: x + 1, lambda x: x**2])

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

ax1.plot(x_cont, y_cont, 'b-', linewidth=2, label='f(x) = x²')
ax1.set_title('Fungsi Kontinu', fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.legend()

ax2.plot(x_cont[x_cont < 0], y_discont[x_cont < 0], 'b-', linewidth=2, label='f(x) = x+1 untuk x<0')
ax2.plot(x_cont[x_cont >= 0], y_discont[x_cont >= 0], 'r-', linewidth=2, label='f(x) = x² untuk x≥0')
ax2.plot(0, 1, 'bo', markersize=8, label='(0,1) dari kiri')
ax2.plot(0, 0, 'ro', markersize=8, label='(0,0) dari kanan')
ax2.set_title('Fungsi Diskontinu di x=0', fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.legend()

plt.tight_layout()
plt.show()
````

## 2. Turunan (Derivatives) 📈

### Definisi Turunan
Turunan adalah **laju perubahan sesaat** dari suatu fungsi terhadap variabelnya. Secara geometris, turunan merepresentasikan **kemiringan garis singgung** pada suatu titik di kurva.

**Definisi formal:**
$$f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}$$

### Notasi Turunan
- $f'(x)$ atau $\frac{df}{dx}$ atau $\frac{dy}{dx}$
- Turunan kedua: $f''(x)$ atau $\frac{d^2f}{dx^2}$

### Aturan Dasar Turunan
1. **Aturan Konstanta**: $\frac{d}{dx}(c) = 0$
2. **Aturan Pangkat**: $\frac{d}{dx}(x^n) = nx^{n-1}$
3. **Aturan Jumlah**: $\frac{d}{dx}[f(x) + g(x)] = f'(x) + g'(x)$
4. **Aturan Perkalian**: $\frac{d}{dx}[f(x)g(x)] = f'(x)g(x) + f(x)g'(x)$
5. **Aturan Pembagian**: $\frac{d}{dx}\left[\frac{f(x)}{g(x)}\right] = \frac{f'(x)g(x) - f(x)g'(x)}{[g(x)]^2}$
6. **Aturan Rantai**: $\frac{d}{dx}[f(g(x))] = f'(g(x)) \cdot g'(x)$

### Turunan Fungsi Trigonometri
- $\frac{d}{dx}(\sin x) = \cos x$
- $\frac{d}{dx}(\cos x) = -\sin x$
- $\frac{d}{dx}(\tan x) = \sec^2 x$

### Turunan Fungsi Eksponensial dan Logaritma
- $\frac{d}{dx}(e^x) = e^x$
- $\frac{d}{dx}(a^x) = a^x \ln a$
- $\frac{d}{dx}(\ln x) = \frac{1}{x}$
- $\frac{d}{dx}(\log_a x) = \frac{1}{x \ln a}$

**Referensi:**
- [Khan Academy - Introduction to Derivatives](https://www.khanacademy.org/math/calculus-1/cs1-derivatives)
- [MIT OCW - Derivatives](https://ocw.mit.edu/courses/mathematics/18-01sc-single-variable-calculus-fall-2010/)

````python
# Implementasi Turunan dengan SymPy
import sympy as sp
import matplotlib.pyplot as plt

# Definisikan variabel simbolik
x = sp.Symbol('x')

# Contoh fungsi
f = x**3 - 2*x**2 + x + 1
print(f"Fungsi f(x) = {f}")

# Hitung turunan pertama
f_prime = sp.diff(f, x)
print(f"Turunan pertama f'(x) = {f_prime}")

# Hitung turunan kedua
f_double_prime = sp.diff(f, x, 2)
print(f"Turunan kedua f''(x) = {f_double_prime}")

# Contoh turunan dengan aturan rantai
g = sp.sin(x**2 + 1)
g_prime = sp.diff(g, x)
print(f"\nFungsi g(x) = {g}")
print(f"Turunan g'(x) = {g_prime}")

# Contoh turunan implisit
# x^2 + y^2 = 25 (lingkaran)
y = sp.Function('y')
equation = x**2 + y(x)**2 - 25
implicit_derivative = sp.solve(sp.diff(equation, x), sp.diff(y(x), x))[0]
print(f"\nTurunan implisit dy/dx untuk x² + y² = 25:")
print(f"dy/dx = {implicit_derivative}")

# Visualisasi fungsi dan turunannya
import numpy as np

# Konversi ke fungsi numerik
f_numeric = sp.lambdify(x, f, 'numpy')
f_prime_numeric = sp.lambdify(x, f_prime, 'numpy')

# Buat data untuk plot
x_vals = np.linspace(-3, 4, 1000)
y_vals = f_numeric(x_vals)
y_prime_vals = f_prime_numeric(x_vals)

# Plot fungsi dan turunannya
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.plot(x_vals, y_vals, 'b-', linewidth=2, label='f(x)')
plt.grid(True, alpha=0.3)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Fungsi f(x) = x³ - 2x² + x + 1')
plt.legend()

plt.subplot(2, 2, 2)
plt.plot(x_vals, y_prime_vals, 'r-', linewidth=2, label="f'(x)")
plt.axhline(y=0, color='k', linestyle='--', alpha=0.5)
plt.grid(True, alpha=0.3)
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title('Turunan f\'(x) = 3x² - 4x + 1')
plt.legend()

# Plot kedua fungsi bersama-sama
plt.subplot(2, 2, 3)
plt.plot(x_vals, y_vals, 'b-', linewidth=2, label='f(x)')
plt.plot(x_vals, y_prime_vals, 'r-', linewidth=2, label="f'(x)")
plt.axhline(y=0, color='k', linestyle='--', alpha=0.5)
plt.axvline(x=0, color='k', linestyle='--', alpha=0.5)
plt.grid(True, alpha=0.3)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Fungsi dan Turunannya')
plt.legend()

# Visualisasi garis singgung di titik tertentu
plt.subplot(2, 2, 4)
x0 = 1  # Titik di mana kita akan menggambar garis singgung
y0 = f_numeric(x0)
slope = f_prime_numeric(x0)

# Fungsi dan turunan
plt.plot(x_vals, y_vals, 'b-', linewidth=2, label='f(x)')

# Titik singgung
plt.plot(x0, y0, 'ro', markersize=8, label=f'Titik ({x0}, {y0:.1f})')

# Garis singgung: y - y0 = m(x - x0)
tangent_x = np.linspace(x0-1, x0+1, 100)
tangent_y = y0 + slope * (tangent_x - x0)
plt.plot(tangent_x, tangent_y, 'g--', linewidth=2, label=f'Garis singgung (slope={slope:.1f})')

plt.grid(True, alpha=0.3)
plt.xlabel('x')
plt.ylabel('y')
plt.title(f'Garis Singgung di x = {x0}')
plt.legend()
plt.xlim(x0-1.5, x0+1.5)
plt.ylim(y0-2, y0+2)

plt.tight_layout()
plt.show()

# Aplikasi: Mencari titik kritis (maksimum dan minimum)
critical_points = sp.solve(f_prime, x)
print(f"\nTitik kritis (f'(x) = 0): {critical_points}")

for point in critical_points:
    second_derivative_value = f_double_prime.subs(x, point)
    print(f"Pada x = {point}:")
    print(f"  f''({point}) = {second_derivative_value}")
    if second_derivative_value > 0:
        print(f"  → Minimum lokal")
    elif second_derivative_value < 0:
        print(f"  → Maksimum lokal")
    else:
        print(f"  → Uji lebih lanjut diperlukan")
````

## 3. Integral 📏

### Definisi Integral
Integral adalah **kebalikan dari turunan** (antiderivative). Secara geometris, integral tentu merepresentasikan **luas daerah di bawah kurva**.

### Integral Tak Tentu (Indefinite Integral)
$$\int f(x) \, dx = F(x) + C$$
dimana $F'(x) = f(x)$ dan $C$ adalah konstanta integrasi.

### Integral Tentu (Definite Integral)
$$\int_a^b f(x) \, dx = F(b) - F(a)$$

### Teorema Fundamental Kalkulus
1. **Bagian 1**: Jika $F(x) = \int_a^x f(t) \, dt$, maka $F'(x) = f(x)$
2. **Bagian 2**: $\int_a^b f(x) \, dx = F(b) - F(a)$ dimana $F'(x) = f(x)$

### Aturan Dasar Integral
1. **Aturan Konstanta**: $\int c \, dx = cx + C$
2. **Aturan Pangkat**: $\int x^n \, dx = \frac{x^{n+1}}{n+1} + C$ (untuk $n \neq -1$)
3. **Aturan Jumlah**: $\int [f(x) + g(x)] \, dx = \int f(x) \, dx + \int g(x) \, dx$
4. **Aturan Konstanta Keluar**: $\int c \cdot f(x) \, dx = c \int f(x) \, dx$

### Integral Fungsi Dasar
- $\int \frac{1}{x} \, dx = \ln|x| + C$
- $\int e^x \, dx = e^x + C$
- $\int a^x \, dx = \frac{a^x}{\ln a} + C$
- $\int \sin x \, dx = -\cos x + C$
- $\int \cos x \, dx = \sin x + C$
- $\int \sec^2 x \, dx = \tan x + C$

### Teknik Integrasi
1. **Substitusi**: $\int f(g(x))g'(x) \, dx = \int f(u) \, du$ dimana $u = g(x)$
2. **Parsial**: $\int u \, dv = uv - \int v \, du$
3. **Pecahan Parsial**: Untuk fungsi rasional

**Referensi:**
- [Khan Academy - Introduction to Integration](https://www.khanacademy.org/math/calculus-1/cs1-integration-and-ftc)
- [Paul's Online Math Notes - Integration](https://tutorial.math.lamar.edu/classes/calci/integrals.aspx)

````python
# Implementasi Integral dengan SymPy
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate

# Definisikan variabel simbolik
x = sp.Symbol('x')

# Contoh 1: Integral tak tentu
f1 = x**2 + 3*x + 2
integral_f1 = sp.integrate(f1, x)
print(f"Fungsi f(x) = {f1}")
print(f"Integral tak tentu ∫f(x)dx = {integral_f1} + C")

# Contoh 2: Integral tentu
a, b = 0, 3
definite_integral = sp.integrate(f1, (x, a, b))
print(f"\nIntegral tentu ∫₀³ f(x)dx = {definite_integral}")

# Contoh 3: Integral dengan substitusi
f2 = sp.sin(2*x)
integral_f2 = sp.integrate(f2, x)
print(f"\nFungsi g(x) = {f2}")
print(f"Integral ∫sin(2x)dx = {integral_f2} + C")

# Contoh 4: Integral parsial
f3 = x * sp.exp(x)
integral_f3 = sp.integrate(f3, x)
print(f"\nFungsi h(x) = {f3}")
print(f"Integral ∫x·e^x dx = {integral_f3} + C")

# Visualisasi integral sebagai luas daerah
plt.figure(figsize=(15, 10))

# Plot 1: Visualisasi luas daerah di bawah kurva
plt.subplot(2, 3, 1)
x_vals = np.linspace(-1, 4, 1000)
f1_numeric = sp.lambdify(x, f1, 'numpy')
y_vals = f1_numeric(x_vals)

plt.plot(x_vals, y_vals, 'b-', linewidth=2, label='f(x) = x² + 3x + 2')

# Area di bawah kurva dari 0 ke 3
x_fill = np.linspace(0, 3, 300)
y_fill = f1_numeric(x_fill)
plt.fill_between(x_fill, 0, y_fill, alpha=0.3, color='lightblue', label=f'Luas = {float(definite_integral):.2f}')

plt.axhline(y=0, color='k', linestyle='-', alpha=0.3)
plt.axvline(x=0, color='k', linestyle='-', alpha=0.3)
plt.axvline(x=3, color='r', linestyle='--', alpha=0.7)
plt.grid(True, alpha=0.3)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Integral Tentu sebagai Luas Daerah')
plt.legend()
plt.xlim(-0.5, 3.5)

# Plot 2: Perbandingan fungsi dan antiturunannya
plt.subplot(2, 3, 2)
antiderivative_numeric = sp.lambdify(x, integral_f1, 'numpy')
y_antiderivative = antiderivative_numeric(x_vals)

plt.plot(x_vals, y_vals, 'b-', linewidth=2, label='f(x)')
plt.plot(x_vals, y_antiderivative, 'r-', linewidth=2, label='F(x) = ∫f(x)dx')
plt.grid(True, alpha=0.3)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Fungsi dan Antiturunannya')
plt.legend()

# Plot 3: Integral numerik vs simbolik
plt.subplot(2, 3, 3)
def f_numeric(x):
    return x**2 + 3*x + 2

# Metode trapezoid
x_trap = np.linspace(0, 3, 50)
y_trap = f_numeric(x_trap)
area_trapz = np.trapz(y_trap, x_trap)

# Metode Simpson
area_simpson, _ = integrate.quad(f_numeric, 0, 3)

plt.plot(x_vals, y_vals, 'b-', linewidth=2, label='f(x)')
plt.fill_between(x_fill, 0, y_fill, alpha=0.2, color='blue', label=f'Simbolik: {float(definite_integral):.4f}')

# Visualisasi trapezoid
for i in range(len(x_trap)-1):
    plt.plot([x_trap[i], x_trap[i]], [0, y_trap[i]], 'r-', alpha=0.3)
    plt.plot([x_trap[i+1], x_trap[i+1]], [0, y_trap[i+1]], 'r-', alpha=0.3)
    plt.plot([x_trap[i], x_trap[i+1]], [y_trap[i], y_trap[i+1]], 'r-', alpha=0.3)

plt.title(f'Metode Numerik\nTrapezoid: {area_trapz:.4f}\nSimpson: {area_simpson:.4f}')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot 4: Integral fungsi trigonometri
plt.subplot(2, 3, 4)
x_trig = np.linspace(0, 2*np.pi, 1000)
y_sin = np.sin(x_trig)
y_cos = np.cos(x_trig)

plt.plot(x_trig, y_sin, 'b-', linewidth=2, label='sin(x)')
plt.plot(x_trig, -y_cos, 'r-', linewidth=2, label='-cos(x) = ∫sin(x)dx')
plt.axhline(y=0, color='k', linestyle='-', alpha=0.3)
plt.grid(True, alpha=0.3)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Integral Fungsi Trigonometri')
plt.legend()

# Plot 5: Fungsi eksponensial dan integralnya
plt.subplot(2, 3, 5)
x_exp = np.linspace(-2, 2, 1000)
y_exp = np.exp(x_exp)

plt.plot(x_exp, y_exp, 'b-', linewidth=2, label='e^x')
plt.plot(x_exp, y_exp, 'r--', linewidth=2, label='∫e^x dx = e^x + C')
plt.grid(True, alpha=0.3)
plt.xlabel('x')
plt.ylabel('y')
plt.title('e^x adalah Antiturunan dari Dirinya Sendiri')
plt.legend()

# Plot 6: Teorema Fundamental Kalkulus
plt.subplot(2, 3, 6)
def F(t):
    """F(t) = ∫₀ᵗ (x² + 1) dx"""
    result, _ = integrate.quad(lambda x: x**2 + 1, 0, t)
    return result

def f_original(x):
    return x**2 + 1

t_vals = np.linspace(0, 3, 100)
F_vals = [F(t) for t in t_vals]

plt.subplot(2, 3, 6)
plt.plot(t_vals, F_vals, 'b-', linewidth=2, label='F(t) = ∫₀ᵗ (x² + 1) dx')

# Turunan F(t) harus sama dengan f(t)
F_derivative = np.gradient(F_vals, t_vals)
plt.plot(t_vals, f_original(t_vals), 'r--', linewidth=2, label='f(t) = t² + 1')
plt.plot(t_vals, F_derivative, 'g:', linewidth=2, alpha=0.7, label="F'(t) ≈ f(t)")

plt.grid(True, alpha=0.3)
plt.xlabel('t')
plt.ylabel('y')
plt.title('Teorema Fundamental Kalkulus')
plt.legend()

plt.tight_layout()
plt.show()

# Contoh aplikasi integral: Menghitung volume benda putar
print("\n" + "="*50)
print("APLIKASI INTEGRAL: VOLUME BENDA PUTAR")
print("="*50)

# Volume benda putar dengan metode cakram
# Fungsi f(x) = √x diputar mengelilingi sumbu x dari x=0 ke x=4
f_volume = sp.sqrt(x)
volume_disk = sp.pi * sp.integrate(f_volume**2, (x, 0, 4))
print(f"Volume benda putar f(x) = √x (0 ≤ x ≤ 4) = {volume_disk}")

# Volume dengan metode kulit tabung
# Volume = 2π ∫ x·f(x) dx
volume_shell = 2 * sp.pi * sp.integrate(x * f_volume, (x, 0, 4))
print(f"Volume dengan metode kulit tabung = {volume_shell}")

# Menghitung panjang kurva
# L = ∫ √(1 + (dy/dx)²) dx
f_curve = x**2
f_curve_prime = sp.diff(f_curve, x)
arc_length_integrand = sp.sqrt(1 + f_curve_prime**2)
# Integral ini kompleks, kita hitung numerik
from scipy.integrate import quad

def arc_length_func(x):
    return np.sqrt(1 + (2*x)**2)

arc_length_numeric, _ = quad(arc_length_func, 0, 2)
print(f"Panjang kurva y = x² dari x=0 ke x=2 ≈ {arc_length_numeric:.4f}")

print("\n" + "="*50)
````

## 4. Aplikasi Kalkulus dalam Kehidupan Nyata 🌍

### 1. Optimisasi (Maksimum dan Minimum)
Kalkulus sangat berguna untuk mencari nilai **maksimum** dan **minimum** dalam berbagai masalah praktis.

**Langkah-langkah optimisasi:**
1. Tentukan fungsi yang akan dioptimalkan
2. Cari turunan pertama $f'(x) = 0$
3. Tentukan titik kritis
4. Gunakan uji turunan kedua untuk menentukan maksimum/minimum
5. Periksa nilai di batas domain

### 2. Analisis Kurva
- **Naik/Turun**: $f'(x) > 0$ (naik), $f'(x) < 0$ (turun)
- **Cekung ke atas/bawah**: $f''(x) > 0$ (cekung ke atas), $f''(x) < 0$ (cekung ke bawah)
- **Titik belok**: $f''(x) = 0$ dan tanda $f''(x)$ berubah

### 3. Laju Perubahan
Turunan merepresentasikan **laju perubahan sesaat**:
- Kecepatan: $v(t) = \frac{ds}{dt}$
- Percepatan: $a(t) = \frac{dv}{dt} = \frac{d²s}{dt²}$
- Laju pertumbuhan populasi, ekonomi, dll.

### 4. Aplikasi Integral
- **Luas daerah** di bawah/antara kurva
- **Volume benda putar**
- **Panjang kurva**
- **Pusat massa** dan momen
- **Kerja** yang dilakukan gaya

- [Khan Academy - Applications of Derivatives](https://www.khanacademy.org/math/calculus-1/cs1-derivatives-applied)

- [Wolfram MathWorld - Calculus](http://mathworld.wolfram.com/Calculus.html)

````python
# Aplikasi Kalkulus: Masalah Optimisasi Nyata
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize_scalar, minimize

print("🚀 APLIKASI KALKULUS DALAM KEHIDUPAN NYATA")
print("="*60)

# PROBLEM 1: Optimisasi Kotak (Masalah Klasik)
print("\n📦 PROBLEM 1: OPTIMISASI KOTAK")
print("-" * 40)

x = sp.Symbol('x', positive=True)

# Misalkan kita punya kertas persegi 20x20 cm
# Kita ingin membuat kotak dengan memotong persegi di setiap sudut
# Berapa ukuran potongan agar volume maksimum?

L = 20  # Panjang sisi kertas
# Volume kotak = panjang × lebar × tinggi
# V(x) = (L-2x) × (L-2x) × x
V = (L - 2*x)**2 * x
print(f"Fungsi volume: V(x) = {V}")

# Cari turunan pertama
dV_dx = sp.diff(V, x)
print(f"Turunan pertama: dV/dx = {dV_dx}")

# Cari titik kritis
critical_points = sp.solve(dV_dx, x)
print(f"Titik kritis: {critical_points}")

# Filter untuk nilai yang masuk akal (0 < x < L/2)
valid_points = [point for point in critical_points if point.is_real and 0 < point < L/2]
print(f"Titik kritis valid: {valid_points}")

# Uji turunan kedua
d2V_dx2 = sp.diff(dV_dx, x)
for point in valid_points:
    second_deriv_value = d2V_dx2.subs(x, point)
    volume_value = V.subs(x, point)
    print(f"Pada x = {point:.3f}:")
    print(f"  Volume = {float(volume_value):.2f} cm³")
    print(f"  d²V/dx² = {float(second_deriv_value):.2f}")
    if second_deriv_value < 0:
        print(f"  → Maksimum lokal ✓")

# Visualisasi
x_vals = np.linspace(0.1, 9.9, 1000)
V_numeric = sp.lambdify(x, V, 'numpy')
V_vals = V_numeric(x_vals)

plt.figure(figsize=(15, 12))

plt.subplot(2, 3, 1)
plt.plot(x_vals, V_vals, 'b-', linewidth=2)
optimal_x = float(valid_points[0])
optimal_V = float(V.subs(x, optimal_x))
plt.plot(optimal_x, optimal_V, 'ro', markersize=10, label=f'Maksimum: x={optimal_x:.2f}, V={optimal_V:.1f}')
plt.xlabel('x (cm)')
plt.ylabel('Volume (cm³)')
plt.title('Optimisasi Volume Kotak')
plt.grid(True, alpha=0.3)
plt.legend()

# PROBLEM 2: Optimisasi Bisnis (Keuntungan Maksimum)
print(f"\n💰 PROBLEM 2: OPTIMISASI KEUNTUNGAN BISNIS")
print("-" * 50)

# Fungsi permintaan: p = 100 - 0.5q (harga vs quantity)
# Fungsi biaya: C = 20q + 1000 (biaya tetap + biaya variabel)
q = sp.Symbol('q', positive=True)
p = 100 - 0.5*q  # Harga per unit
C = 20*q + 1000   # Total biaya
R = p * q         # Revenue = harga × quantity
profit = R - C    # Keuntungan = revenue - cost

print(f"Fungsi permintaan: p = {p}")
print(f"Fungsi biaya: C = {C}")
print(f"Fungsi revenue: R = {R}")
print(f"Fungsi profit: π = {profit}")

# Optimisasi keuntungan
dprofit_dq = sp.diff(profit, q)
optimal_q = sp.solve(dprofit_dq, q)[0]
optimal_p = p.subs(q, optimal_q)
optimal_profit = profit.subs(q, optimal_q)

print(f"\nOptimal quantity: {optimal_q} unit")
print(f"Optimal price: ${float(optimal_p):.2f}")
print(f"Maximum profit: ${float(optimal_profit):.2f}")

# Visualisasi
q_vals = np.linspace(0, 200, 1000)
profit_numeric = sp.lambdify(q, profit, 'numpy')
revenue_numeric = sp.lambdify(q, R, 'numpy')
cost_numeric = sp.lambdify(q, C, 'numpy')

profit_vals = profit_numeric(q_vals)
revenue_vals = revenue_numeric(q_vals)
cost_vals = cost_numeric(q_vals)

plt.subplot(2, 3, 2)
plt.plot(q_vals, profit_vals, 'g-', linewidth=2, label='Profit')
plt.plot(q_vals, revenue_vals, 'b-', linewidth=2, label='Revenue')
plt.plot(q_vals, cost_vals, 'r-', linewidth=2, label='Cost')
plt.axvline(x=float(optimal_q), color='orange', linestyle='--', alpha=0.7, label=f'Optimal q={optimal_q}')
plt.axhline(y=0, color='k', linestyle='-', alpha=0.3)
plt.xlabel('Quantity')
plt.ylabel('Money ($)')
plt.title('Optimisasi Keuntungan Bisnis')
plt.legend()
plt.grid(True, alpha=0.3)

# PROBLEM 3: Laju Perubahan (Fisika)
print(f"\n🏃‍♂️ PROBLEM 3: ANALISIS GERAK")
print("-" * 40)

t = sp.Symbol('t')
# Fungsi posisi: s(t) = t³ - 6t² + 9t + 2
s = t**3 - 6*t**2 + 9*t + 2
v = sp.diff(s, t)  # Kecepatan = ds/dt
a = sp.diff(v, t)  # Percepatan = dv/dt

print(f"Posisi: s(t) = {s}")
print(f"Kecepatan: v(t) = {v}")
print(f"Percepatan: a(t) = {a}")

# Kapan benda berhenti? (v = 0)
stop_times = sp.solve(v, t)
print(f"Benda berhenti pada t = {stop_times}")

# Kapan percepatan = 0?
zero_accel_time = sp.solve(a, t)[0]
print(f"Percepatan = 0 pada t = {zero_accel_time}")

# Visualisasi gerak
t_vals = np.linspace(0, 5, 1000)
s_numeric = sp.lambdify(t, s, 'numpy')
v_numeric = sp.lambdify(t, v, 'numpy')
a_numeric = sp.lambdify(t, a, 'numpy')

s_vals = s_numeric(t_vals)
v_vals = v_numeric(t_vals)
a_vals = a_numeric(t_vals)

plt.subplot(2, 3, 3)
plt.plot(t_vals, s_vals, 'b-', linewidth=2, label='Posisi s(t)')
plt.plot(t_vals, v_vals, 'r-', linewidth=2, label='Kecepatan v(t)')
plt.plot(t_vals, a_vals, 'g-', linewidth=2, label='Percepatan a(t)')
plt.axhline(y=0, color='k', linestyle='-', alpha=0.3)
for stop_time in stop_times:
    if 0 <= float(stop_time) <= 5:
        plt.axvline(x=float(stop_time), color='orange', linestyle='--', alpha=0.7)
plt.xlabel('Waktu (t)')
plt.ylabel('Nilai')
plt.title('Analisis Gerak: Posisi, Kecepatan, Percepatan')
plt.legend()
plt.grid(True, alpha=0.3)

# PROBLEM 4: Optimisasi dengan Kendala (Lagrange)
print(f"\n🎯 PROBLEM 4: OPTIMISASI DENGAN KENDALA")
print("-" * 50)

# Minimisasi f(x,y) = x² + y² dengan kendala g(x,y) = x + y - 1 = 0
# Menggunakan metode Lagrange multiplier
x, y, lam = sp.symbols('x y lambda')

f = x**2 + y**2        # Fungsi objektif
g = x + y - 1          # Kendala

# Gradien dan kondisi Lagrange
grad_f = [sp.diff(f, x), sp.diff(f, y)]
grad_g = [sp.diff(g, x), sp.diff(g, y)]

# Kondisi: ∇f = λ∇g dan g = 0
equations = [
    grad_f[0] - lam * grad_g[0],  # ∂f/∂x = λ∂g/∂x
    grad_f[1] - lam * grad_g[1],  # ∂f/∂y = λ∂g/∂y
    g                             # Kendala g = 0
]

solution = sp.solve(equations, [x, y, lam])
print(f"Solusi optimisasi dengan kendala:")
print(f"x = {solution[x]}, y = {solution[y]}")
print(f"Nilai minimum f = {f.subs(solution)}")

# PROBLEM 5: Integral untuk Menghitung Luas Tidak Beraturan
print(f"\n📐 PROBLEM 5: LUAS DAERAH DENGAN INTEGRAL")
print("-" * 50)

# Menghitung luas daerah antara dua kurva
# y₁ = x² dan y₂ = 2x + 3 dari x = -1 ke x = 3
x = sp.Symbol('x')
y1 = x**2
y2 = 2*x + 3

# Titik potong
intersection_points = sp.solve(y1 - y2, x)
print(f"Kurva berpotongan di x = {intersection_points}")

# Luas = ∫|y₂ - y₁| dx
# Kita perlu cek mana yang di atas
area_total = 0
segments = [-1] + intersection_points + [3]

for i in range(len(segments)-1):
    a, b = segments[i], segments[i+1]
    # Cek fungsi mana yang di atas di interval ini
    mid_point = (a + b) / 2
    if (y2 - y1).subs(x, mid_point) > 0:
        area_segment = sp.integrate(y2 - y1, (x, a, b))
    else:
        area_segment = sp.integrate(y1 - y2, (x, a, b))
    area_total += area_segment
    print(f"Luas dari x={a} ke x={b}: {float(area_segment):.3f}")

print(f"Total luas: {float(area_total):.3f}")

# Visualisasi
x_vals = np.linspace(-1.5, 3.5, 1000)
y1_vals = x_vals**2
y2_vals = 2*x_vals + 3

plt.subplot(2, 3, 4)
plt.plot(x_vals, y1_vals, 'b-', linewidth=2, label='y = x²')
plt.plot(x_vals, y2_vals, 'r-', linewidth=2, label='y = 2x + 3')

# Fill area antara kurva
x_fill = np.linspace(-1, 3, 300)
y1_fill = x_fill**2
y2_fill = 2*x_fill + 3
plt.fill_between(x_fill, y1_fill, y2_fill, alpha=0.3, color='green', 
                label=f'Luas = {float(area_total):.2f}')

# Titik potong
for point in intersection_points:
    y_intersect = float(y1.subs(x, point))
    plt.plot(float(point), y_intersect, 'ko', markersize=8)

plt.axhline(y=0, color='k', linestyle='-', alpha=0.3)
plt.axvline(x=0, color='k', linestyle='-', alpha=0.3)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Luas Daerah Antara Dua Kurva')
plt.legend()
plt.grid(True, alpha=0.3)

# PROBLEM 6: Template untuk Masalah Optimisasi Umum
plt.subplot(2, 3, 5)
plt.text(0.1, 0.9, "📝 TEMPLATE OPTIMISASI", fontsize=14, fontweight='bold', transform=plt.gca().transAxes)
template_text = """
1. Definisikan fungsi objektif f(x)
2. Tentukan domain/kendala
3. Cari f'(x) = 0 (titik kritis)
4. Uji turunan kedua:
   • f''(x) > 0 → minimum
   • f''(x) < 0 → maksimum
5. Periksa nilai di batas domain
6. Bandingkan semua kandidat
7. Pilih nilai optimal

Contoh Kode:
import sympy as sp
x = sp.Symbol('x')
f = ... # fungsi objektif
f_prime = sp.diff(f, x)
critical_pts = sp.solve(f_prime, x)
"""
plt.text(0.05, 0.05, template_text, fontsize=10, transform=plt.gca().transAxes, 
         verticalalignment='bottom', fontfamily='monospace')
plt.axis('off')

# Summary plot
plt.subplot(2, 3, 6)
applications = ['Optimisasi\nKotak', 'Keuntungan\nBisnis', 'Analisis\nGerak', 
                'Kendala\nLagrange', 'Luas\nDaerah']
techniques = ['Turunan\nPertama', 'Titik\nKritis', 'Kecepatan\nPercepatan', 
              'Multiplier\nLagrange', 'Integral\nTentu']

y_pos = np.arange(len(applications))
plt.barh(y_pos, [1, 1, 1, 1, 1], color=['skyblue', 'lightgreen', 'orange', 'pink', 'lightcoral'])
plt.yticks(y_pos, applications)
plt.xlabel('Aplikasi Kalkulus')
plt.title('Ringkasan Aplikasi')
plt.grid(True, alpha=0.3, axis='x')

plt.tight_layout()
plt.show()

print("\n" + "="*60)
print("✅ KESIMPULAN: Kalkulus memiliki aplikasi luas dalam:")
print("• Optimisasi (bisnis, teknik, ekonomi)")
print("• Analisis gerak (fisika, teknik)")
print("• Menghitung luas, volume, panjang kurva")
print("• Pemodelan laju perubahan")
print("• Pemecahan masalah dengan kendala")
print("="*60)
````

### Template Aplikasi Optimisasi
```python
# LANGKAH OPTIMISASI:
# 1. Definisikan variabel dan fungsi objektif
# 2. Tentukan kendala (jika ada)
# 3. Cari turunan dan set = 0
# 4. Solve untuk titik kritis
# 5. Uji turunan kedua
# 6. Evaluasi di batas domain
# 7. Pilih nilai optimal

def solve_optimization_problem():
    # Step 1: Define variables and objective function
    x = sp.Symbol('x', positive=True)  # Add constraints as needed
    objective_function = ...  # Your function to optimize
    
    # Step 2: Find derivative
    derivative = sp.diff(objective_function, x)
    
    # Step 3: Solve for critical points
    critical_points = sp.solve(derivative, x)
    
    # Step 4: Second derivative test
    second_derivative = sp.diff(derivative, x)
    
    for point in critical_points:
        if point.is_real:  # Only consider real solutions
            second_deriv_value = second_derivative.subs(x, point)
            objective_value = objective_function.subs(x, point)
            
            print(f"At x = {point}:")
            print(f"  Objective value = {objective_value}")
            print(f"  Second derivative = {second_deriv_value}")
            
            if second_deriv_value > 0:
                print("  → Local minimum")
            elif second_deriv_value < 0:
                print("  → Local maximum")
    
    return critical_points
```

## 🎯 Kesimpulan dan Rangkuman Kalkulus

### Konsep Utama yang Telah Dipelajari:

1. **Limit** 📍
   - Dasar untuk memahami kontinuitas
   - Foundasi untuk turunan dan integral
   - Teknik evaluasi limit

2. **Turunan** 📈
   - Laju perubahan sesaat
   - Kemiringan garis singgung
   - Aplikasi dalam optimisasi

3. **Integral** 📏
   - Antiturunan (kebalikan turunan)
   - Luas di bawah kurva
   - Volume dan aplikasi geometri

4. **Aplikasi Praktis** 🌍
   - Optimisasi bisnis dan teknik
   - Analisis gerak dalam fisika
   - Pemecahan masalah nyata

### Hubungan Antar Konsep:
$$\text{Limit} \rightarrow \text{Turunan} \rightarrow \text{Aplikasi Optimisasi}$$
$$\text{Limit} \rightarrow \text{Integral} \rightarrow \text{Aplikasi Geometri}$$

### Teorema Fundamental Kalkulus:
Menghubungkan turunan dan integral sebagai operasi kebalikan:
$$\frac{d}{dx} \int_a^x f(t) \, dt = f(x)$$
